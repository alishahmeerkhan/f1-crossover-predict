```python
import os
import pandas as pd
import seaborn as sns
from dotenv import load_dotenv
import matplotlib.pyplot as plt
```

```python
load_dotenv(override=True)

file_path = os.getenv('LOAD_CLEAN_PATH')

if not file_path:
    raise ValueError('CRITICAL: File path not found.')
```

```python
df = pd.read_csv(file_path)
```

```python
df.head()
```

## Data Details
To predict -> LapTime_s<br>
How much time will be lost (time delta) give the current situation (features)

```python
fig, axis = plt.subplots(1, 2, figsize=(15, 7))
plt.subplot(121)
sns.barplot(data=df, x='Compound', y='time_delta', hue='Rainfall')

plt.subplot(122)
sns.scatterplot(data=df, x='TrackTemp', y='time_delta')
plt.show()
```

```python
fig, axis = plt.subplots(1, 3, figsize=(18, 5))

sns.lineplot(data=df, x='TyreLife', y='time_delta', ax=axis[0])
axis[0].set_title('Tyre Degradation')


sns.lineplot(data=df, x='AirTemp', y='time_delta', ax=axis[1])
axis[1].set_title('Air Temp Vs. Time Delta (s)')


sns.lineplot(data=df, x='TrackTemp', y='time_delta', ax=axis[2])
axis[2].set_title('Track Temp Vs. Time Delta (s)')

plt.tight_layout()
plt.show()
```
